# Create Admissions Table

```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
icu_file_path = '/home/lu.kuang/thesis/Prospective_DataCleaning/updated_cleaning/DataMerged_Prospective_0430_updated.xlsx'
```

```python
ADMISSIONS = pd.read_excel(io = icu_file_path, sheet_name = 'PATIENT VISIT')
ADMISSIONS.columns
```

```python
# keeping only important fields
ADMISSIONS = ADMISSIONS[['RECORD_ID','MRN','HOSP_ADMIT','HOSP_DISCHARGE','ICU_ADMIT','ICU_DISCHARGE','SITE',
                     'ADMITTED_FROM','ADMIT_TYPE','ADMIT_CLASS','ADMIT_DX', 'ICU_SURVIVAL',
                     'HOSP_SURVIVAL', 'ICU_LOS', 'HOSP_LOS',]]
ADMISSIONS
```

```python
PATIENTS = pd.read_csv('PATIENTS.csv', sep=";", encoding='Latin-1', index_col="ROW_ID")
PATIENTS
```

```python
ADMISSIONS = ADMISSIONS.merge(right = PATIENTS[['MRN','PATIENT_ID','DOD']], on = ['MRN'])
# Convert relevant columns to datetime
datetime_cols = ['HOSP_ADMIT', 'HOSP_DISCHARGE', 'ICU_ADMIT', 'ICU_DISCHARGE']

for col in datetime_cols:
    ADMISSIONS[col] = pd.to_datetime(ADMISSIONS[col], errors='coerce')
ADMISSIONS
```

```python
# adjusting names of fields
ADMISSIONS.rename(columns={'HOSP_ADMIT':'HOSP_ADMIT_DATETIME',
                           'HOSP_DISCHARGE':'HOSP_DISCH_DATETIME',
                           'ICU_ADMIT':'ICU_ADMIT_DATETIME',
                           'ICU_DISCHARGE':'ICU_DISCH_DATETIME',
                           'SITE':'ADMISSION_SITE',
                           'UNIT_ADMITTED_FROM':'ADMITTED_FROM',
                           'ADMIT_TYPE':'ADMISSION_TYPE',
                           'ADMIT_CLASS':'ADMISSION_CLASS',
                           'ADMIT_DX':'ADMISSION_DIAGNOSIS',
}, inplace = True)
ADMISSIONS
```

```python
# renaming index to ROW_ID
ADMISSIONS.index.set_names(['ROW_ID'], inplace = True)
```

```python
# adjusting RECORD_ID to int
ADMISSIONS['RECORD_ID'] = ADMISSIONS['RECORD_ID'].apply(lambda x: int(x))
```

```python
# creating ADMISSION_ID
ADMISSIONS['ADMISSION_ID'] = range(0,len(ADMISSIONS))
ADMISSIONS['ADMISSION_ID'] = ADMISSIONS['ADMISSION_ID'].apply(lambda x: 'PA'+str(x).rjust(5, '0'))
```

```python
# create HOSP_READMIT
patient_ids = sorted(list(set(ADMISSIONS['PATIENT_ID'])))
for patient_id in patient_ids:
    temp = ADMISSIONS[ADMISSIONS['PATIENT_ID'] == patient_id].sort_values(by = ['HOSP_ADMIT_DATETIME'], ascending = True).reset_index()
    for x in range(len(temp)):
        index = temp.loc[x,'ROW_ID']
        if x == 0:
            temp.loc[x,'HOSP_READMIT'] = 0
        else:
            if temp.loc[x,'HOSP_ADMIT_DATETIME'] == temp.loc[x-1,'HOSP_ADMIT_DATETIME']:
                temp.loc[x,'HOSP_READMIT'] = temp.loc[x-1,'HOSP_READMIT']
            else:
                temp.loc[x,'HOSP_READMIT'] = temp.loc[x-1,'HOSP_READMIT'] + 1
        
        ADMISSIONS.loc[index,'HOSP_READMIT'] = temp.loc[x,'HOSP_READMIT']
```

```python
# create RECUR_ICU_ADMIT
patient_ids = sorted(list(set(ADMISSIONS['PATIENT_ID'])))
for patient_id in patient_ids:
    admissions = len(set(ADMISSIONS[ADMISSIONS['PATIENT_ID'] == patient_id]['HOSP_READMIT']))    
    for admission in range(admissions):
        temp = ADMISSIONS[(ADMISSIONS['PATIENT_ID'] == patient_id) & 
                          (ADMISSIONS['HOSP_READMIT'] == admission)].sort_values(by = ['ICU_ADMIT_DATETIME'], ascending = True).reset_index()
        for x in range(len(temp)):
            index = temp.loc[x,'ROW_ID']
            if x == 0:
                temp.loc[x,'ICU_READMIT'] = 0
            else:
                if temp.loc[x,'ICU_ADMIT_DATETIME'] == temp.loc[x-1,'ICU_ADMIT_DATETIME']:
                    temp.loc[x,'ICU_READMIT'] = temp.loc[x-1,'ICU_READMIT']
                else:
                    temp.loc[x,'ICU_READMIT'] = temp.loc[x-1,'ICU_READMIT'] + 1

            ADMISSIONS.loc[index,'ICU_READMIT'] = temp.loc[x,'ICU_READMIT']
```

```python
# Map survival status to binary mortality indicators
survival_map = {'A': 0, 'D': 1}

ADMISSIONS['ICU_MORTALITY'] = ADMISSIONS['ICU_SURVIVAL'].map(survival_map)
ADMISSIONS['HOSP_MORTALITY'] = ADMISSIONS['HOSP_SURVIVAL'].map(survival_map)
ADMISSIONS
```

```python
ADMISSIONS.to_csv('ADMISSIONS.csv', sep = ";", encoding = 'Latin-1')
```
